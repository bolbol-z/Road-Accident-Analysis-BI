```python
import pandas as pd
import numpy as np
```

```python
# 1. CHARGEMENT DU FICHIER
# ==========================================

caract = pd.read_csv(r"C:\Users\AOURUS\Desktop\isgis\3eme\S1\python BI\projet\dataset\caract-2023.csv", sep=";")

print("\n=== 1. APERÇU DES DONNÉES ===")
print(caract.head())
print("\nDimensions :", caract.shape)
```

```python
# 1. Affichage des infos (Informations de base sur le DataFrame)
print("\n--- INFOS / TYPES DE DONNEES (Structure et Types) ---")
caract.info()
```

```python
# 2. Affichage des Valeurs manquantes
print("\n--- VALEURS MANQUANTES (Comptage) ---")
# Afficher le total des valeurs manquantes pour chaque colonne
print(caract.isnull().sum())
# Ajouter l'affichage du pourcentage des valeurs manquantes pour plus de clarté
print("\n--- POURCENTAGE DES VALEURS MANQUANTES ---")
print((caract.isnull().sum() / len(caract)) * 100)
```

```python
# 3. Affichage du nombre de doublons
print("\n--- DOUBLONS (Nombre de lignes dupliquees) ---")
# Calculer le nombre total de lignes entièrement dupliquées
print(caract.duplicated().sum())
```

```python
# 4. ANALYSE STATISTIQUE
# ==========================================

print("\n=== 4. ANALYSE STATISTIQUE ===")
print(caract.describe().T) 
```

```python
# 5. Format des chaînes
# ====================================================================
print("\n--- FORMAT DES CHAINES (Verification des espaces et caracteres invisibles) ---")
# Sélectionner uniquement les colonnes de type 'object' (chaîne de caractères)
object_cols = caract.select_dtypes(include=['object']).columns

for col in object_cols:
    # Utiliser str.strip() pour compter les valeurs avec des espaces blancs excédentaires (début/fin)
    count_leading_trailing_spaces = caract[col].astype(str).apply(lambda x: x.strip() != x).sum()
    print(f"[{col}]: Nombre de valeurs avec des espaces excédentaires (Leading/Trailing Spaces): {count_leading_trailing_spaces}")
    if count_leading_trailing_spaces > 0:
        print(f"  > Exemple d'une valeur avec espace excédentaire : {caract[col][caract[col].astype(str).apply(lambda x: x.strip() != x)].iloc[0]!r}")
```

```python
# 6. Normalisation des catégories
# ====================================================================
print("\n--- NORMALISATION DES CATEGORIES (Coherence des niveaux) ---")
# Afficher le nombre de catégories uniques (Unique Categories) et leurs valeurs les plus fréquentes (Top 5) pour chaque colonne catégorielle.
for col in object_cols:
    print(f"\n[{col}]: {caract[col].nunique()} catégories uniques")
    # Afficher les 5 valeurs les plus fréquentes (pour détecter les variations d'orthographe comme "Manual" et "manuel")
    print(caract[col].value_counts().head(5))
```

```python
# 7. Nom des colonnes
# ====================================================================
print("\n--- NOM DES COLONNES (Noms des colonnes) ---")
# Afficher tous les noms de colonnes pour vérifier le format (Exemple : espaces ou majuscules/minuscules)
print(caract.columns.tolist())
```

```python
# 8. Uniformisation des unités / formats
# ====================================================================
print("\n--- UNIFORMISATION DES UNITES / FORMATS (Formatage des nombres et des dates) ---")
print("Note: Cette vérification dépend des colonnes. Utilisez 'describe().T' et 'info()' pour vous assurer que les nombres ne sont pas stockés comme texte ('object').")
print("Colonnes numériques actuelles:")
print(caract.select_dtypes(include=np.number).columns.tolist())
```

```python
# -----------9. DÉTECTION DE PROBLÈMES -----------

print("\n=== 9. PROBLÈMES DÉTECTÉS ===")
```

```python
# b) Coordonnées lat/long avec virgules
bad_lat = caract["lat"].astype(str).str.contains(',')
bad_long = caract["long"].astype(str).str.contains(',')

if bad_lat.any() or bad_long.any():
    print("- Problème: lat/long utilisent la virgule au lieu du point.")

# c) Horaires hrmn en texte
if caract["hrmn"].dtype == object:
    print("- Problème: hrmn est au format texte au lieu de datetime.")

```

```python
# 1. Uniformisation des noms des colonnes
# ===========================
caract.columns = caract.columns.str.lower().str.strip()

# Vérifier à nouveau (après normalisation)
print("Colonnes après normalisation:", caract.columns.tolist())
```

```python
#Remplacer adresse vide par "Inconnue"
caract["adr"] = caract["adr"].fillna("Adresse_inconnue")
```

```python
print("\n--- VALEURS MANQUANTES APRES NETTOYAGE ---")
print(caract.isnull().sum())
```

```python
# 2. Nettoyage des chaînes de caractères dans la colonne "adr"
# Retirer espaces inutiles
caract["adr"] = caract["adr"].str.strip()

# Mettre tout en minuscule ou majuscule
caract["adr"] = caract["adr"].str.lower()   # ou .str.upper()

# Remettre première lettre en majuscule
caract["adr"] = caract["adr"].str.title()
#Supprimer accents (é → e, ç → c, etc.)
import unidecode
caract["adr"] = caract["adr"].apply(lambda x: unidecode.unidecode(x) if isinstance(x, str) else x)

```

```python
# --- Convertir hrmn en format HH:MM ---
caract["hrmn"] = pd.to_datetime(caract["hrmn"], errors="coerce")
print("✔ hrmn converti au format datetime.")
```

```python
# --- Convertir lat/long en float ---
caract["lat"] = caract["lat"].astype(str).str.replace(",", ".", regex=False).astype(float)
caract["long"] = caract["long"].astype(str).str.replace(",", ".", regex=False).astype(float)
print("✔ lat/long convertis en float.")
```

```python
print("\n=== 1. APERÇU DES DONNÉES ===")
print(caract.head())
print("\nDimensions :", caract.shape)
```

```python

# Supprime les espaces en début et fin sur la colonne 'adr'
caract['adr'] = caract['adr'].str.strip()
```

```python
# 9. Sauvegarde du fichier nettoyé au format CSV (UTF-8)
# ===========================
OUTPUT_PATH = "caracts_nettoye.csv"
caract.to_csv(OUTPUT_PATH, index=False, encoding="utf-8")
print(f"\n✅ Le fichier nettoyé a été sauvegardé sous: {OUTPUT_PATH}")
```

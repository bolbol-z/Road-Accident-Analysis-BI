```python
import pandas as pd
import numpy as np
```

```python
# 1. CHARGEMENT DU FICHIER
# ==========================================

lieux = pd.read_csv(r"C:\Users\AOURUS\Desktop\isgis\3eme\S1\python BI\projet\dataset\lieux-2023.csv", sep=";")

print("\n=== 1. APERÇU DES DONNÉES ===")
print(lieux.head())
print("\nDimensions :", lieux.shape)
```

```python
# 1. Affichage des infos (Informations de base sur le DataFrame)
print("\n--- INFOS / TYPES DE DONNEES (Structure et Types) ---")
lieux.info()
```

```python
# 2. Affichage des Valeurs manquantes
print("\n--- VALEURS MANQUANTES (Comptage) ---")
# Afficher le total des valeurs manquantes pour chaque colonne
print(lieux.isnull().sum())
# Ajouter l'affichage du pourcentage des valeurs manquantes pour plus de clarté
print("\n--- POURCENTAGE DES VALEURS MANQUANTES ---")
print((lieux.isnull().sum() / len(lieux)) * 100)
```

```python
# 3. Affichage du nombre de doublons
print("\n--- DOUBLONS (Nombre de lignes dupliquees) ---")
# Calculer le nombre total de lignes entièrement dupliquées
print(lieux.duplicated().sum())
```

```python
# 4. ANALYSE STATISTIQUE
# ==========================================

print("\n=== 4. ANALYSE STATISTIQUE ===")
print(lieux.describe().T) 
```

```python
# 5. Format des chaînes
# ====================================================================
print("\n--- FORMAT DES CHAINES (Verification des espaces et caracteres invisibles) ---")
# Sélectionner uniquement les colonnes de type 'object' (chaîne de caractères)
object_cols = lieux.select_dtypes(include=['object']).columns

for col in object_cols:
    # Utiliser str.strip() pour compter les valeurs avec des espaces blancs excédentaires (début/fin)
    count_leading_trailing_spaces = lieux[col].astype(str).apply(lambda x: x.strip() != x).sum()
    print(f"[{col}]: Nombre de valeurs avec des espaces excédentaires (Leading/Trailing Spaces): {count_leading_trailing_spaces}")
    if count_leading_trailing_spaces > 0:
        print(f"  > Exemple d'une valeur avec espace excédentaire : {lieux[col][lieux[col].astype(str).apply(lambda x: x.strip() != x)].iloc[0]!r}")
```

```python
# 6. Normalisation des catégories
# ====================================================================
print("\n--- NORMALISATION DES CATEGORIES (Coherence des niveaux) ---")
# Afficher le nombre de catégories uniques (Unique Categories) et leurs valeurs les plus fréquentes (Top 5) pour chaque colonne catégorielle.
for col in object_cols:
    print(f"\n[{col}]: {lieux[col].nunique()} catégories uniques")
    # Afficher les 5 valeurs les plus fréquentes (pour détecter les variations d'orthographe comme "Manual" et "manuel")
    print(lieux[col].value_counts().head(5))
```

```python
# 7. Nom des colonnes
# ====================================================================
print("\n--- NOM DES COLONNES (Noms des colonnes) ---")
# Afficher tous les noms de colonnes pour vérifier le format (Exemple : espaces ou majuscules/minuscules)
print(lieux.columns.tolist())
```

```python
# 1. Uniformisation des noms des colonnes
# ===========================
lieux.columns = lieux.columns.str.lower().str.strip()

# Vérifier à nouveau (après normalisation)
print("Colonnes après normalisation:", lieux.columns.tolist())
```

```python
#Remplacer voix vide par "Inconnue"
lieux["voie"] = lieux["voie"].fillna("voie_inconnue")
```

```python
# 2. Nettoyage des chaînes de caractères dans la colonne "voie"
# Retirer espaces inutiles
lieux["voie"] = lieux["voie"].str.strip()

# Mettre tout en minuscule ou majuscule
lieux["voie"] = lieux["voie"].str.lower()   # ou .str.upper()

# Remettre première lettre en majuscule
lieux["voie"] = lieux["voie"].str.title()
#Supprimer accents (é → e, ç → c, etc.)
import unidecode
lieux["voie"] = lieux["voie"].apply(lambda x: unidecode.unidecode(x) if isinstance(x, str) else x)

```

```python
# Supprime les espaces en début et fin sur la colonne 'voie' et 'v2
lieux['voie'] = lieux['voie'].str.strip()
lieux['v2'] = lieux['v2'].str.strip()
```

```python
# 5. Format des chaînes
# ====================================================================
print("\n--- FORMAT DES CHAINES (Verification des espaces et caracteres invisibles) ---")
# Sélectionner uniquement les colonnes de type 'object' (chaîne de caractères)
object_cols = lieux.select_dtypes(include=['object']).columns

for col in object_cols:
    # Utiliser str.strip() pour compter les valeurs avec des espaces blancs excédentaires (début/fin)
    count_leading_trailing_spaces = lieux[col].astype(str).apply(lambda x: x.strip() != x).sum()
    print(f"[{col}]: Nombre de valeurs avec des espaces excédentaires (Leading/Trailing Spaces): {count_leading_trailing_spaces}")
    if count_leading_trailing_spaces > 0:
        print(f"  > Exemple d'une valeur avec espace excédentaire : {lieux[col][lieux[col].astype(str).apply(lambda x: x.strip() != x)].iloc[0]!r}")
```

```python
# 2. Affichage des Valeurs manquantes
print("\n--- VALEURS MANQUANTES (Comptage) ---")
# Afficher le total des valeurs manquantes pour chaque colonne
print(lieux.isnull().sum())
# Ajouter l'affichage du pourcentage des valeurs manquantes pour plus de clarté
print("\n--- POURCENTAGE DES VALEURS MANQUANTES ---")
print((lieux.isnull().sum() / len(lieux)) * 100)
```

```python
# Supprimer les colonnes 'v2' et 'lartpc'
lieux = lieux.drop(columns=['v2', 'lartpc'])
```

```python
# 2. Affichage des Valeurs manquantes
print("\n--- VALEURS MANQUANTES (Comptage) ---")
# Afficher le total des valeurs manquantes pour chaque colonne
print(lieux.isnull().sum())
# Ajouter l'affichage du pourcentage des valeurs manquantes pour plus de clarté
print("\n--- POURCENTAGE DES VALEURS MANQUANTES ---")
print((lieux.isnull().sum() / len(lieux)) * 100)
```

```python
# 9. Sauvegarde du fichier nettoyé au format CSV (UTF-8)
# ===========================
OUTPUT_PATH = "lieux_nettoye.csv"
lieux.to_csv(OUTPUT_PATH, index=False, encoding="utf-8")
print(f"\n✅ Le fichier nettoyé a été sauvegardé sous: {OUTPUT_PATH}")
```

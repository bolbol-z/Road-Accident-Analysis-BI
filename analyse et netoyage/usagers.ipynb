```python
import pandas as pd
import numpy as np
```

```python
# 1. CHARGEMENT DU FICHIER
# ==========================================

usagers = pd.read_csv(r"C:\Users\AOURUS\Desktop\isgis\3eme\S1\python BI\projet\dataset\usagers-2023.csv", sep=";")

print("\n=== 1. APERÇU DES DONNÉES ===")
print(usagers.head())
print("\nDimensions :", usagers.shape)
```

```python
# 1. Affichage des infos (Informations de base sur le DataFrame)
print("\n--- INFOS / TYPES DE DONNEES (Structure et Types) ---")
usagers.info()
```

```python
# 2. Affichage des Valeurs manquantes
print("\n--- VALEURS MANQUANTES (Comptage) ---")
# Afficher le total des valeurs manquantes pour chaque colonne
print(usagers.isnull().sum())
# Ajouter l'affichage du pourcentage des valeurs manquantes pour plus de clarté
print("\n--- POURCENTAGE DES VALEURS MANQUANTES ---")
print((usagers.isnull().sum() / len(usagers)) * 100)
```

```python
# 3. Affichage du nombre de doublons
print("\n--- DOUBLONS (Nombre de lignes dupliquees) ---")
# Calculer le nombre total de lignes entièrement dupliquées
print(usagers.duplicated().sum())
```

```python
# 4. ANALYSE STATISTIQUE
# ==========================================

print("\n=== 4. ANALYSE STATISTIQUE ===")
print(usagers.describe().T) 
```

```python
# 5. Format des chaînes
# ====================================================================
print("\n--- FORMAT DES CHAINES (Verification des espaces et caracteres invisibles) ---")
# Sélectionner uniquement les colonnes de type 'object' (chaîne de caractères)
object_cols = usagers.select_dtypes(include=['object']).columns

for col in object_cols:
    # Utiliser str.strip() pour compter les valeurs avec des espaces blancs excédentaires (début/fin)
    count_leading_trailing_spaces = usagers[col].astype(str).apply(lambda x: x.strip() != x).sum()
    print(f"[{col}]: Nombre de valeurs avec des espaces excédentaires (Leading/Trailing Spaces): {count_leading_trailing_spaces}")
    if count_leading_trailing_spaces > 0:
        print(f"  > Exemple d'une valeur avec espace excédentaire : {usagers[col][usagers[col].astype(str).apply(lambda x: x.strip() != x)].iloc[0]!r}")
```

```python
# 6. Normalisation des catégories
# ====================================================================
print("\n--- NORMALISATION DES CATEGORIES (Coherence des niveaux) ---")
# Afficher le nombre de catégories uniques (Unique Categories) et leurs valeurs les plus fréquentes (Top 5) pour chaque colonne catégorielle.
for col in object_cols:
    print(f"\n[{col}]: {usagers[col].nunique()} catégories uniques")
    # Afficher les 5 valeurs les plus fréquentes (pour détecter les variations d'orthographe comme "Manual" et "manuel")
    print(usagers[col].value_counts())
```

```python
# 8. Uniformisation des unités / formats
# ====================================================================
print("\n--- UNIFORMISATION DES UNITES / FORMATS (Formatage des nombres et des dates) ---")
print("Note: Cette vérification dépend des colonnes. Utilisez 'describe().T' et 'info()' pour vous assurer que les nombres ne sont pas stockés comme texte ('object').")
print("Colonnes numériques actuelles:")
print(usagers.select_dtypes(include=np.number).columns.tolist())
```

```python
# 5. Détection valeurs -1 (souvent = inconnu dans ce dataset)
colonnes_code = ["secu1", "secu2", "secu3", "locp", "actp", "etatp"]
for c in colonnes_code:
    print(f"\nValeurs -1 dans {c} :", (usagers[c] == -1).sum())
```

```python
print("\n=== REPARTITION GRAVITE ===")
print(usagers["grav"].value_counts())
```

```python
# 1. Uniformisation des noms des colonnes
# ===========================
usagers.columns = usagers.columns.str.lower().str.strip()

# Vérifier à nouveau (après normalisation)
print("Colonnes après normalisation:", usagers.columns.tolist())
```

```python
# 1) Remplacer les valeurs manquantes de 'an_nais' par sa médiane
median_an_nais = usagers["an_nais"].median()
usagers["an_nais"] = usagers["an_nais"].fillna(median_an_nais)
```

```python
print("\n=== AJOUT DE LA COLONNE AGE ===")
usagers["age"] = 2023 - usagers["an_nais"]
```

```python
# 2. Affichage des Valeurs manquantes
print("\n--- VALEURS MANQUANTES (Comptage) ---")
# Afficher le total des valeurs manquantes pour chaque colonne
print(usagers.isnull().sum())
# Ajouter l'affichage du pourcentage des valeurs manquantes pour plus de clarté
print("\n--- POURCENTAGE DES VALEURS MANQUANTES ---")
print((usagers.isnull().sum() / len(usagers)) * 100)
```

```python
print("\n=== 1. APERÇU DES DONNÉES ===")
print(usagers.head())
print("\nDimensions :", usagers.shape)
```

```python
# Remplacer les valeurs -1 par la valeur maximale  dans la colonne 'grav'
usagers["grav"] = usagers["grav"].replace(-1, 1)

# Vérifier la nouvelle répartition
print("=== REPARTITION GRAVITE APRES REMPLACEMENT ===")
print(usagers["grav"].value_counts().sort_index())
```

```python
# 9. Sauvegarde du fichier nettoyé au format CSV (UTF-8)
# ===========================
OUTPUT_PATH = "usagers_nettoye.csv"
usagers.to_csv(OUTPUT_PATH, index=False, encoding="utf-8")
print(f"\n✅ Le fichier nettoyé a été sauvegardé sous: {OUTPUT_PATH}")
```
